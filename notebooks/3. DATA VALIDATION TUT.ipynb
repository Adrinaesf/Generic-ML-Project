## 2. Info of Missing data:

```python
# 2.1 Find the missing values:
df.isna() 
```

```python
# 2.2: Count the missing values in each colunm:
df.isna().sum()
```

```python
# 2.3 percenatage missing in each colunm:
((df.isna().sum()) / len(df)) * 100
```

```python
# 2.4 Handeling the missing data:
# Remove rows with missing values:
df.dropna()
# Remove columns with missing values
df.dropna(axis=1)
# Fill with the mean/median in a column:
df["math score"] = df["math score"].fillna(
    df["math score"].mean() # .median()
)
# Fill with most common value
df["gender"] = df["gender"].fillna(
    df["gender"].mode()[0]
)
```
