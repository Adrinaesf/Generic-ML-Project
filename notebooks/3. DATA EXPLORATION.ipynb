## 1. Understanding the data
1. Load the data
2. Looking the first and last 5 row
3. Shape (rows, cols) 
4. Data types of the data in columns (features)
5. Unique Values

```python
# 1.1 Load the data
df = pd.read_csv("data/stud.csv")
```

```python
# 1.2 Looking the first and last 5 row
print("First 5 rows:")
print(df.head())
print()
print()
print("Last 5 rows:")
print(df.tail())
```

```python
# 1.3 Shape of data: (rows, cols)
df.shape # output: (1000, 8)
```

```python
# 1.4: Features (cols name)
df.columns
```

```python
# 1.5 Data types
df.info()
```

```python
df.describe() # Statics summary of integer data types. 
```

```python
# 1.6 Unique Values: 
print("Unique values of a categorical data: ")
print(df["race/ethnicity"].unique())
print()
print("Number of unique values:")
print(df["race/ethnicity"].nunique())
```

```python
# 1.7: Count each category
print("Number of people in different grups of ethnicity:")
print(df["race/ethnicity"].value_counts())
```

## 2. Info of Missing data:
1. Find the missing value
2. Counting the number of missing values
3. Percentage missing

```python
# 2.1 Find the missing values:
df.isna() 
```

```python
# 2.2: Count the missing values in each colunm:
df.isna().sum()
```

```python
# 2.3 percenatage missing in each colunm:
((df.isna().sum()) / len(df)) * 100
```

## 3. Handelling the Missing Data
1. Remove rows with missing values
2. Remove columns with missing values
3. Filling with mean and median
4. Fill with most common value

```python
# 3.1 Remove rows with missing values:
df.dropna()

# 3.2 Remove columns with missing values
df.dropna(axis=1)

# 3.3 Fill with the mean/median in a column:
df["math score"] = df["math score"].fillna(
    df["math score"].mean() # .median()
)

# 3.4 Fill with most common value
df["gender"] = df["gender"].fillna(
    df["gender"].mode()[0]
)
```

## 4. Info of Duplicates
1. Count the number of duplicates
2. Remove duplicates

```python
# 4.1 Count the duplicates
print(df.duplicated().sum())

# 4.2 Remove the duplicate
df.drop_duplicates()
```

## 5. Selecting Data
1. Getting one or more culumn
2. Getting first n rows
3. Getting a specific row
4. GEtting a range of rows

```python
# 5.1 Getting one or more column:
df["math score"] # one column
df[["math score", "reading score"]] # mult column

# 5.2 Getting the first n rows: (n = 3)
df.head(3)

# 5.3 Getting a specific row:
df.iloc[167] # getting row 167th

# 5.4 Getting a range of rows: 
df.iloc[160:168] # row 160-168
```

## 6. Filtering the data
1. Getting the data with a spicific condition
2. See examples

```python
# 6.1 students with math score > 80:
df[df["math score"] > 80]

# 6.2 Female students:
df[df["gender"] == "female"]

# 6.3 students with math AND reading > 80:
df[
    (df["math score"] > 80)
    & 
    (df["reading score"] > 80)
]
# Note: either condition would be: | (means or)
```

## 7. Creating New Columns
1. Make your own coulmn using the otehrs
2. See examples

```python
# 7.1 Make your own column:
df["average"] = (df["math score"] + df["reading score"] + df["writing score"]) / 3

# 7.2 pass or fail:
df["passed"] = df["average"] >= 50
```

## 8. Sorting
1. Sorting a spicific colunm
    * `sort_values(factor, ascending=T/F)` will return a new DataFrame and it won't modify the previous one, so to see the result we have to create a new DF and assign its values to that. 
2. Y/N ascedning

```python
# 8.1 Sorting a specific column:
first_11_df = df.iloc[0:11].copy() # making the data smaller for faster computation
sorted_df_asc = first_11_df.sort_values("math score") 
sorted_df_desc = first_11_df.sort_values("math score", ascending=False) 

# print("Low to High")
# print(sorted_df["math score"])
# print()
# print("High to Low")
print(sorted_df_desc["math score"])
```

## 9. Aggregations
1. Average/mean
2. Max/Min
3. Median
4. Standard deviation

```python
# 9.1: Average:
print("Information of math grades:")
print("Average: {} \nMax: {} \nMin: {} \nMedian: {} \nSTD: {}".format(
    df["math score"].mean(), 
    df["math score"].max(), 
    df["math score"].min(), 
    df["math score"].median(), 
    df["math score"].std(), 
    ))
```

## 10. Grouping + Correlation 

```python
# VERY IMPORTANT:
# 10.1 Average math score by gender
print(df.groupby("gender")["math score"].mean())

print()
# 10.2 grouping the min reading score by race
print(df.groupby("race/ethnicity")["reading score"].min())

print()
# Correlation: Useful for seeing how strongly numerical variables are related.
print(df.corr(numeric_only=True))
```
