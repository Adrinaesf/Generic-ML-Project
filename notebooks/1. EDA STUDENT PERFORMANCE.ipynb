### 3.7 define numerical & categorical columns

```python
# define numerical & categorical columns
numeric_features = [feature for feature in df.columns if df[feature].dtype == 'int64']
categorical_features = [feature for feature in df.columns if df[feature].dtype != 'int64']

# print columns
print(df["math score"].dtype)
print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

### 3.8 Adding columns for "Total Score" and "Average"

```python
# 3.8 Adding columns for "Total Score" and "Average"
df["Total Score"] = df["math score"] + df["writing score"] + df["reading score"]
df["Average of Scores"] = df["Total Score"] / 3
df.head()
```

```python
reading_full = df[df["reading score"] == 100]["Average of Scores"].count()
math_full = df[df["math score"] == 100]["Average of Scores"].count()
writing_full = df[df["writing score"] == 100]["Average of Scores"].count()

print(f'Number of students with full marks in Maths: {math_full}')
print(f'Number of students with full marks in Writing: {writing_full}')
print(f'Number of students with full marks in Reading: {reading_full}')
```

```python
reading_less_20 = df[df['reading score'] <= 20]['Average of Scores'].count()
writing_less_20 = df[df['writing score'] <= 20]['Average of Scores'].count()
math_less_20 = df[df['math score'] <= 20]['Average of Scores'].count()

print(f'Number of students with less than 20 marks in Maths: {math_less_20}')
print(f'Number of students with less than 20 marks in Writing: {writing_less_20}')
print(f'Number of students with less than 20 marks in Reading: {reading_less_20}')
```

**Insights**
* From above values we get students have performed the worst in Maths
* Best performance is in reading section

# 4) **Data Visualization**

### 4.1 Visualize average and total score distribution
* Histogram
* Kernel Distribution Function (KDE)

```python
fig, axes = plt.subplots(
    nrows = 1, 
    ncols = 2, 
    figsize = (12,6)
)

sns.histplot(
    data = df, 
    x = "Average of Scores", 
    bins=30, 
    kde=True, 
    color='g', 
    ax = axes[0]
)

sns.histplot(
    data = df, 
    x = "Average of Scores", 
    bins=30, 
    kde=True, 
    hue="gender",
    ax = axes[1]
)
```

```python
fig, axes = plt.subplots(
    nrows = 1, 
    ncols = 2, 
    figsize = (12,6)
)

sns.histplot(
    data = df, 
    x = "Total Score", 
    bins=30, 
    kde=True, 
    color='g', 
    ax = axes[0]
)

sns.histplot(
    data = df, 
    x = "Total Score", 
    bins=30, 
    kde=True, 
    hue="gender",
    ax = axes[1]
)
```

**Insights:**

* Female students tent to perform well than male students
